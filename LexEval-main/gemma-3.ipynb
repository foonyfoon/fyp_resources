```python
from transformers import AutoTokenizer

# Load the tokenizer for the Gemma 3 IT 4B model
def number_of_token(text):
    tokenizer = AutoTokenizer.from_pretrained("google/gemma-3-12b-it")

    # Tokenize and count tokens
    tokens = tokenizer.tokenize(text)
    token_ids = tokenizer(text)["input_ids"]
    num_tokens = len(token_ids)

    print(f"Tokenized text: {tokens}")
    print(f"Number of tokens: {num_tokens}")

def format_topk_wiki_answer(document_list: list) -> str:
    extracts = "\n".join(f"{doc['title']}: {doc['content']}" for doc in document_list)
    extracts = extracts.replace("{", "").replace("}", "")

    # Build the final text prompt
    text_template = (
        "You are a helpful and honest assistant. Your answers should not include any harmful, unethical, racist, sexist, toxic, dangerous, "
        "or illegal content. You have retrieved the following extracts from the Wikipedia pages:\n{extracts}\n"
        "You are expected to give truthful and concise answers based on the previous extracts. If it doesn't include relevant information "
        "for the request just say so and don't make up false information.\n"
        "Keep the answers as concise as possible"
        # ", does not have to be full sentences. For example, for the question: "
        # "What is Taylor Swift's occupation? Your response should be: Singer, Singer-songwriter"
    )

    # Format the text with the combined extracts
    text = text_template.format(extracts=extracts)
    return text

def bfs_token_count(tree):
    queue = list(tree.root.children)  # Skip the root

    while queue:
        node = queue.pop(0)
        queue.extend(node.children)

        rag_closest_match = node.rag_closest_match
        rag_closest_passage = [match[0] for match in rag_closest_match]

        input_text = format_topk_wiki_answer(rag_closest_passage)
        tokens = number_of_token(input_text)
        print(tokens)
```

```python
import pandas as pd
import re

def parse_step_timings(log_text: str) -> pd.DataFrame:
    pattern = r"Q:\s*(\d+),\s*Model:.*?,\s*Step:\s*([\w_]+)\s*->\s*([\d.]+)s"
    matches = re.findall(pattern, log_text)

    data = [{
        "query": int(query),
        "step": step,
        "time": float(time)
    } for query, step, time in matches]

    return pd.DataFrame(data)

def calculate_average_step_times(df: pd.DataFrame) -> pd.DataFrame:
    return df.groupby("step")["time"].mean().reset_index().sort_values(by="time", ascending=False)
```
