```python
import os
import re

dir = '/vol/bitbucket/lst20/logs'
```

```python

# Define the pattern to match
pattern = re.compile(r"INFO - Question of index: \d+ cannot be processed, adding to DLQ\.")

# Directory containing the files (change if needed)
directory = "."

# Find all files that start with "02-06"
files_to_process = [f for f in os.listdir(directory) if f.startswith(f"{dir}/02-06")]

# Set to store unique lines
unique_lines = set()

# Process each file
for file_name in files_to_process:
    file_path = os.path.join(directory, file_name)
    with open(file_path, "r", encoding="utf-8") as file:
        for line in file:
            if pattern.match(line.strip()):
                unique_lines.add(line.strip())  # Add to set to ensure uniqueness

# Output file to save results
output_file = "processed_dlq_info.log"

# Write unique lines to the output file
with open(output_file, "w", encoding="utf-8") as output:
    output.write("\n".join(sorted(unique_lines)))

print(f"Processed {len(files_to_process)} files.")
print(f"Extracted {len(unique_lines)} unique matching line")
```
