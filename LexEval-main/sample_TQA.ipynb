```python
size_ratio = 1000 / len(labelled_df)
# Split the data to get a subset with the same distribution of classes
labelled_df["original_index"] = labelled_df.index
_, df_shuffled, _, _ = train_test_split(
    labelled_df,
    labelled_df["cluster_id"],
    test_size=size_ratio,
    shuffle=True,
    random_state=constants.SEED,
    stratify=labelled_df["cluster_id"],
)
# Ensure the directory exists
out_dir = os.path.dirname(constants.SHUFFLED_TQA_FILE)
if not os.path.exists(out_dir):
    os.makedirs(out_dir)
# Save shuffled dataset
df_shuffled.to_parquet(constants.SHUFFLED_TQA_FILE, index=False)
```

```python
import pandas as pd

# File paths
POPQA_DF_LOCATION = "/vol/bitbucket/lst20/lex-eval_dataset/PopQA/test.csv"
SHUFFLED_FILE = "/vol/bitbucket/lst20/lex-eval_dataset/PopQA/shuffled.csv"

# Output parquet paths (same directory, with .parquet extension)
POPQA_PARQUET = POPQA_DF_LOCATION.replace('.csv', '.parquet')
SHUFFLED_PARQUET = SHUFFLED_FILE.replace('.csv', '.parquet')


def convert_csv_to_parquet(csv_path: str, parquet_path: str):
    """
    Reads a CSV file at `csv_path` into a DataFrame and writes it out as a parquet file.
    """
    print(f"Reading CSV: {csv_path}")
    df = pd.read_csv(csv_path)
    print(f"Saving Parquet: {parquet_path}")
    df.to_parquet(parquet_path, index=False)
    print("Done!\n")


def main():
    # Convert PopQA test set
    convert_csv_to_parquet(POPQA_DF_LOCATION, POPQA_PARQUET)

    # Convert shuffled dataset
    convert_csv_to_parquet(SHUFFLED_FILE, SHUFFLED_PARQUET)
    
main()
```
